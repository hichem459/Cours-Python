```python
# Entry widget : User inputs (textbox that accepts a single line of user input)

import tkinter as tk

def submit():
    #pass
    username = entry.get()
    #print(f'The username is : {username}')
    print("Hello " + username)

def delete():
    entry.delete(0, 'end')   # deletes a line of text 
    
def backspace():
    entry.delete(len(entry.get()) - 1, 'end') # delete last character
    

window = tk.Tk()
window.title("The Entry widget !")

# Add od buttons :
submit = tk.Button(master = window, text = 'submit', command = submit)
submit.pack(side = 'right')

delete = tk.Button(master = window, text = 'delete', command = delete)
delete.pack(side = 'right')

backspace = tk.Button(master = window, text = 'backspace', command = backspace)
backspace.pack(side = 'right')

#entry = tk.Entry(master = window)
entry = tk.Entry()
entry.config(font = ('Ink Free', 50))
entry.config(background = '#111111')
entry.config(foreground = '#6FE5ED')  # #00FF00 light green or Green
#entry.insert(0, 'default text')                     # insert some default text
#entry.config(state = 'disabled') # active/disabled
entry.config(width = 10)
#entry.config(show = '*')  # for password

entry.pack()

window.mainloop()
```

```python
# Check button :

import tkinter as tk

def display():
    #pass
    # y = x.get()
    # print(f"The value is {y}")
    # if (x.get() == 1):
    #     print("I like coding with Python !")
    # else:
    #     print("I don't like coding with Python !")
    if ((x.get() == 1) & (y.get() == 0)):
        print("I like coding with Python !")
    elif ((x.get() == 0) & (y.get() == 1)):
        print("I like coding with Java !")
    elif ((x.get() == 1) & (y.get() == 1)):
        print("I like coding with both Python & Java!")      
    else:
        print("I don't like coding with either !")

window = tk.Tk()
window.title("The check button widget !")

x = tk.IntVar()
y = tk.IntVar()

checkbox = tk.Checkbutton(master = window, 
                          text = 'Python',
                          variable = x,
                          onvalue = 1, 
                          offvalue = 0,
                          command = display)
# Decorating the checkbox
checkbox.config(font = ('Arial', 20))
checkbox.config(foreground = '#0000FF')
checkbox.configure(background = '#000000')
checkbox.config(activebackground = '#000000')
checkbox.config(activeforeground = '#0000FF')

photo = tk.PhotoImage(file = 'python.png')
photo = photo.subsample(25, 25)
checkbox.config(image = photo, compound = 'left')
checkbox.config(padx = 25, pady = 10, width = 250, height = 50)
checkbox.config(anchor = 'w')
checkbox.pack()

checkbox2 = tk.Checkbutton(master = window, 
                          text = 'Java',
                          variable = y,
                          onvalue = 1, 
                          offvalue = 0,
                          command = display)
# Decorating the checkbox
checkbox2.config(font = ('Arial', 20))
checkbox2.config(foreground = '#0000FF')
checkbox2.configure(background = '#000000')
checkbox2.config(activebackground = '#000000')
checkbox2.config(activeforeground = '#0000FF')

photo2 = tk.PhotoImage(file = 'java.png')
photo2 = photo2.subsample(2, 2)
checkbox2.config(image = photo2, compound = 'left')
checkbox2.config(padx = 25, pady = 10, width = 250, height = 50)
checkbox2.config(anchor = 'w')
checkbox2.pack()

window.mainloop()
```

```python
# Radio button : similar to checkbox, but you can only select one from a group

import tkinter as tk


def order():
    if (x.get() == 0):
        print("You ordered pizza !")
    elif (x.get() == 1):
        print("You ordered a hamburger")
    elif (x.get()==2):
        print("You ordered a hotdog")
    else:
        print("Sorry !!")

window = tk.Tk()
window.title("The Radio Button widget !")

food = ["pizza", "hamburger", "hotdog"]

pizzaImage = tk.PhotoImage(file = 'pizza.png')
pizzaImage = pizzaImage.subsample(10, 10)
hamburgerImage = tk.PhotoImage(file = 'hamburger.png')
hamburgerImage = hamburgerImage.subsample(15, 15)
hotdogImage = tk.PhotoImage(file = 'hotdog.png')
hotdogImage= hotdogImage.subsample(10, 10)

foodImages = [pizzaImage, hamburgerImage, hotdogImage]

x = tk.IntVar()

for index in range(len(food)):
    radiobutton = tk.Radiobutton(master = window, 
                                 text = food[index], 
                                 variable = x, 
                                 value = index,
                                 image = foodImages[index],
                                 compound = 'left',
                                 indicatoron = 0,   # eliminate circle indicators
                                 width = 375,       # sets width of radio buttons
                                 command = order    # set command of radiobutton to function
                                 )
    radiobutton.config(padx = 25)
    radiobutton.config(font = ('Impact', 50))
    radiobutton.config()
    radiobutton.pack(anchor = 'w')

window.mainloop()
```

```python
# file dialog : open files

import tkinter as tk
from tkinter import filedialog

def openFile():
    # filepath = filedialog.askopenfilename()
    # #print(filepath)
    # file = open(filepath, 'r')
    # print(file.read())
    # file.close()
    
    filepath = filedialog.askopenfilename(initialdir = "C:\\Users\\ABBASSI\\Desktop\\Python_Tkinter_REVISION",
                                          title = "Open file okay ?",
                                          filetypes = (("text files", "*.text"), ("all files", "*.*"))
                                          )
    if filepath is None:
        return
    file =open(filepath, 'r')
    print(file.read())
    file.close()    

window = tk.Tk()
window.title("The file dialog : opening files !")
window.geometry("400x30")

button = tk.Button(master = window, text = 'Open', command = openFile)
button.pack()

window.mainloop()
```
